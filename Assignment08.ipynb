```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
```

```python
ir = load_iris()
df = pd.DataFrame(ir.data, columns=ir.feature_names)

df.head()
```

```python
print(df.shape)
```

```python
scaler = StandardScaler()
X_Scaled = scaler.fit_transform(df)
print(X_Scaled[:5])
```

K-Means Algorithms
Step 1: Choose K
step 2: Initialize Centroids; Randomly Place K Centrod near the data space
Step 3 : Assign Points
Each point is assigned to nearest centroid
Step 4:Update Centroid
Step 5 : Repeat steps 3 and 4 until centroid stops moving


K-means Clustering algorithm is used in iris dataset because it contains three group of flowers ; here no of clusters = 3

```python
kmeans = KMeans(n_clusters = 3, random_state=42, n_init=10)
kmeans_cl = kmeans.fit_predict(X_Scaled)
df['kmeans_clusters'] = kmeans_cl
df.head()
```

```python
x = X_Scaled[:, 2]
y = X_Scaled[:, 3]
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.scatter(x, y, c=kmeans_cl)

plt.show()
```

Hierarchial Clustering Builds Multilevel Hirearchy of cluster either merging smaller into larger ones(Agglomerative) or dividing larger cluster into smaller ones(Divisive) Agglomerative clustering is suitable for iris dataset to group dataset into 3 group of flowers


```python
hierarchial =AgglomerativeClustering(n_clusters =3)
hierarchial_clus = hierarchial.fit_predict(X_Scaled)
df['hierarchial_clusters'] = hierarchial_clus
df.head()
```

```python
x = X_Scaled[:, 2]
y = X_Scaled[:, 3]
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.scatter(x, y, c=hierarchial_clus)

plt.show()
```
